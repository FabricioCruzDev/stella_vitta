## FORNECEDOR

## PRODUTOS

```python
df_novos_produtos = pd.read_excel('../raw/2- PRODUTOS - PRECIFICAÇAO segunda compra 09.02.xlsx', sheet_name='Sheet1', skiprows=1)
df_novos_produtos
```

```python
df_novos_produtos.columns
```

```python
df_novos_produtos['COD_FORNECEDOR'] = df_novos_produtos['FORNECEDOR'].astype(str).str[-2:]
```

```python
df_novos_produtos['INDEX'] = df_novos_produtos['INDEX'] = range(148, 148 + len(df_novos_produtos))
df_novos_produtos['INDEX'] = df_novos_produtos['INDEX'].apply(lambda x: str(x).zfill(6))
df_novos_produtos['SKU'] = df_novos_produtos['COD_FORNECEDOR'].astype(str) + df_novos_produtos['INDEX'].astype(str) + df_novos_produtos['METAL']
df_novos_produtos_sku = df_novos_produtos[['DESCRICAO', 'SKU']]
# df_novos_produtos_sku.loc[df_novos_produtos_sku['SKU'].isna()]
df_novos_produtos_sku.drop(index=93, inplace=True)
```

```python
df_novos_produtos_sku.to_csv('../output/produtos_sku.csv', index=False, encoding='latin-1')
```
